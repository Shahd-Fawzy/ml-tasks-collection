```python
model  =  tf.keras.Sequential([

                               Embedding(10000 , 128 ,mask_zero=True ) ,
                               SimpleRNN(128 , return_sequences=True , activation="tanh") ,
                               SimpleRNN(64 , return_sequences=False , activation="tanh") ,
                               Dense(1 , activation="sigmoid")
]

)
```

```python
model  =  tf.keras.Sequential([

                               Embedding(10000 , 128 ,mask_zero=True ) ,
                               LSTM(128 , return_sequences=True , activation="tanh") ,
                               LSTM(64 , return_sequences=False , activation="tanh"),
                               Dense(1 , activation="sigmoid")
]

)
```

```python
def encode_review(text  ,word_index):
      words  =  text.lower().split()
      seq =  []
      for word in words :
        seq.append(word_index.get(word , 2))
      return seq
```
